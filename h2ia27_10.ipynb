```python
# Carrega novamente o conjunto de dados original
df = pd.read_csv('https://raw.githubusercontent.com/Rswcf/houseprices_kaggle/master/train.csv')

from sklearn.model_selection import train_test_split

# Divide o dataset em desenvolvimento (80%) e teste (20%)
dev, test = train_test_split(df, test_size=0.2)
# Divide o conjunto de desenvolvimento em treino (80% do dev) e validação (20% do dev)
train, validation = train_test_split(dev, test_size=0.2)

# Calcula a idade da casa para cada um dos três conjuntos de dados
train["HouseAge"] = train["YrSold"] - train["YearBuilt"]
validation["HouseAge"] = validation["YrSold"] - validation["YearBuilt"]
test["HouseAge"] = test["YrSold"] - test["YearBuilt"]

# Define as colunas numéricas e categóricas que serão usadas como características (features) preditoras
numeric_columns = ["GrLivArea", "OverallQual", "GarageCars", "LotFrontage", "HouseAge"]
categorical_columns = ["Neighborhood", "MSSubClass"]
feature_columns = numeric_columns + categorical_columns

# Separa os dados de entrada (X) e o alvo/preço de venda (y) para o conjunto de treino
X_train = train[feature_columns].copy()
y_train = train["SalePrice"]

# Separa as variáveis para o conjunto de validação
X_validation = validation[feature_columns].copy()
y_validation = validation["SalePrice"]

# Separa as variáveis para o conjunto de teste final
X_test = test[feature_columns].copy()
y_test = test["SalePrice"]

# Aplica o One-Hot Encoding (cria colunas booleanas 0/1) nas variáveis categóricas do conjunto de treino
X_train = pd.get_dummies(X_train, columns=categorical_columns, dtype=int)

# Aplica One-Hot Encoding no conjunto de validação e reordena/alinha as colunas de acordo com o treino
X_validation = pd.get_dummies(X_validation, columns=categorical_columns, dtype=int)
X_validation = X_validation.reindex(columns=X_train.columns, fill_value=0)

# Aplica One-Hot Encoding no conjunto de teste e alinha as colunas com o formato esperado pelo treino
X_test = pd.get_dummies(X_test, columns=categorical_columns, dtype=int)
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)

# Calcula a mediana das colunas numéricas no treino para usar no preenchimento de valores ausentes (imputação)
median_values = X_train[numeric_columns].median()

# Preenche quaisquer valores nulos (NaN) nas colunas numéricas de treino, validação e teste usando as medianas de treino
X_train[numeric_columns] = X_train[numeric_columns].fillna(median_values)
X_validation[numeric_columns] = X_validation[numeric_columns].fillna(median_values)
X_test[numeric_columns] = X_test[numeric_columns].fillna(median_values)
```

```python
# Importa a classe de Regressão Linear
from sklearn.linear_model import LinearRegression

# Cria a instância do modelo de Regressão Linear
linear_model = LinearRegression()

# Treina o modelo utilizando os dados de treinamento (variáveis preditoras e preços reais)
linear_model.fit(X_train, y_train)
```

```python
# Importa o modelo de Regressão Linear
from sklearn.linear_model import LinearRegression

# Instancia o algoritmo de regressão linear
linear_model = LinearRegression()

# Treina o modelo com o conjunto de treino
linear_model.fit(X_train, y_train)

# Gera as previsões de preços para os dados do conjunto de validação
linear_predictions = linear_model.predict(X_validation)

# Importa a métrica de Erro Médio Absoluto (MAE)
from sklearn.metrics import mean_absolute_error

# Calcula o erro de referência (baseline) usando a média simples de preço de venda do treino
baseline_error = mean_absolute_error(y_validation, [y_train.mean()] * len(y_validation))

# Calcula o erro absoluto médio das previsões reais geradas pelo modelo de regressão linear
linear_error = mean_absolute_error(y_validation, linear_predictions)

# Exibe os resultados comparativos dos erros médios em dólares (US$)
print("Erro médio da referência (US$):", baseline_error)
print("Erro médio da regressão (US$):", linear_error)
```
